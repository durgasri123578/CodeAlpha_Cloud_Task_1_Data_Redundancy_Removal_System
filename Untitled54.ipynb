# Data Redundancy Removal System

## Task 1

This project finds duplicate data and removes it.

It checks new data before adding it to the database.

Only unique data will be stored.

## Step 1: Create Sample Data

Here, we create a small customer dataset using Python.

```python
import pandas as pd

data = {
    "Customer_ID": [101, 102, 103, 104, 105],
    "Name": ["Arun", "Priya", "Ravi", "Meena", "Kumar"],
    "Email": [
        "arun@gmail.com",
        "priya@gmail.com",
        "ravi@gmail.com",
        "meena@gmail.com",
        "kumar@gmail.com"
    ],
    "City": [
        "Chennai",
        "Madurai",
        "Chengalpattu",
        "Chennai",
        "Kanchipuram"
    ]
}

df = pd.DataFrame(data)

display(df)
```

## Step 2: Find Duplicate Data

Here, we check the dataset for duplicate records.

```python
# Check for duplicate records

duplicates = df[df.duplicated()]

display(duplicates)
```

## Step 3: Add Duplicate Data

Here, we add a duplicate record to the dataset.

This helps us test whether the system can find duplicate data.

```python
# Add a duplicate record

new_data = {
    "Customer_ID": 106,
    "Name": "Arun",
    "Email": "arun@gmail.com",
    "City": "Chennai"
}

df.loc[len(df)] = new_data

display(df)
```

## Step 4: Remove Duplicate Data

Here, we remove the duplicate records from the dataset.

Only one copy of each record will be kept.

```python
# Remove duplicate records

df = df.drop_duplicates(subset=["Name", "Email", "City"])

display(df)
```

## Step 5: Check Unique Data

Here, we check the data after removing duplicates.

This helps us make sure that only unique data is present.

```python
# Check for duplicate records again

duplicates = df[df.duplicated(subset=["Name", "Email", "City"])]

if duplicates.empty:
    print("No duplicate records found.")
else:
    display(duplicates)
```

## Step 6: Validate New Data

Here, we check new data before adding it.

If the data already exists, it will not be added.

```python
# Validate new customer data

new_customer = {
    "Customer_ID": 107,
    "Name": "Priya",
    "Email": "priya@gmail.com",
    "City": "Madurai"
}

if (
    (df["Name"] == new_customer["Name"]) &
    (df["Email"] == new_customer["Email"]) &
    (df["City"] == new_customer["City"])
).any():
    print("Duplicate data found. Data not added.")
else:
    df.loc[len(df)] = new_customer
    print("New data added successfully.")

display(df)
```

## Step 7: Add Unique Data

Here, we add new data that is not already in the dataset.

The system will add the data because it is unique.

```python
# Add a new unique customer

new_customer = {
    "Customer_ID": 107,
    "Name": "Karthik",
    "Email": "karthik@gmail.com",
    "City": "Chennai"
}

if (
    (df["Name"] == new_customer["Name"]) &
    (df["Email"] == new_customer["Email"]) &
    (df["City"] == new_customer["City"])
).any():
    print("Duplicate data found. Data not added.")
else:
    df.loc[len(df)] = new_customer
    print("New data added successfully.")

display(df)
```

## Step 8: Save Clean Data

Here, we save the cleaned data into a CSV file.

The file contains only unique records.

```python
# Save the clean data

df.to_csv("clean_customer_data.csv", index=False)

print("Clean data saved successfully.")
```

## Step 9: Check the Saved File

Here, we check whether the clean data file was created successfully.

```python
# Check the saved file

import os

if os.path.exists("clean_customer_data.csv"):
    print("File created successfully.")
else:
    print("File not found.")
```

## Step 10: Create Database

Here, we create a SQLite database.

The clean customer data will be stored in the database.

```python
import sqlite3

# Create a database
connection = sqlite3.connect("customer_database.db")

# Store the clean data in the database
df.to_sql("customers", connection, if_exists="replace", index=False)

print("Data stored in the database successfully.")
```

## Step 11: Read Data from Database

Here, we read the stored data from the database.

This helps us check whether the data was stored correctly.

```python
# Read data from the database

database_data = pd.read_sql("SELECT * FROM customers", connection)

display(database_data)
```

## Step 12: Final Duplicate Check

Here, we check the database one last time.

There should be no duplicate records.

```python
# Check for duplicate records in the database

duplicate_count = database_data.duplicated(
    subset=["Name", "Email", "City"]
).sum()

if duplicate_count == 0:
    print("No duplicate records found.")
else:
    print("Duplicate records found:", duplicate_count)
```

## Step 13: Close the Database

Here, we close the database connection after completing the work.

```python
# Close the database connection

connection.close()

print("Database connection closed successfully.")
```
